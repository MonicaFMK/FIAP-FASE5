### Convertendo colunas

```python
import pandas as pd
import numpy as np
from datetime import datetime, date

# ============================================
# MAPEAMENTO DAS MÉTRICAS POR ANO
# ============================================

bases_metricas = {
    2020: {
        'df': df_passos_2020,
        'colunas': {
            'IAA': 'IAA_2020',
            'IEG': 'IEG_2020',
            'IPS': 'IPS_2020',
            'IDA': 'IDA_2020',
            'IPP': 'IPP_2020',
            'IPV': 'IPV_2020',
            'IAN': 'IAN_2020'
        }
    },

    2021: {
        'df': df_passos_2021,
        'colunas': {
            'IAA': 'IAA_2021',
            'IEG': 'IEG_2021',
            'IPS': 'IPS_2021',
            'IDA': 'IDA_2021',
            'IPP': 'IPP_2021',
            'IPV': 'IPV_2021',
            'IAN': 'IAN_2021'
        }
    },

    2022: {
        'df': df_passos_2022,
        'colunas': {
            'IAA': 'IAA',
            'IEG': 'IEG',
            'IPS': 'IPS',
            'IDA': 'IDA',
            'IPP': 'IPP',
            'IPV': 'IPV',
            'IAN': 'IAN'
        }
    },

    2023: {
        'df': df_passos_2023,
        'colunas': {
            'IAA': 'IAA',
            'IEG': 'IEG',
            'IPS': 'IPS',
            'IDA': 'IDA',
            'IPP': 'IPP',
            'IPV': 'IPV',
            'IAN': 'IAN'
        }
    },

    2024: {
        'df': df_passos_2024,
        'colunas': {
            'IAA': 'IAA',
            'IEG': 'IEG',
            'IPS': 'IPS',
            'IDA': 'IDA',
            'IPP': 'IPP',
            'IPV': 'IPV',
            'IAN': 'IAN'
        }
    }
}


# ============================================
# FUNÇÃO SOMENTE PARA VERIFICAÇÃO
# NÃO ALTERA OS DADOS
# ============================================

def verificar_metrica(serie):

    qtd_data = 0
    qtd_percentual = 0
    qtd_texto_invalido = 0
    qtd_maior_100 = 0

    exemplos_data = []
    exemplos_percentual = []
    exemplos_texto = []
    exemplos_maior_100 = []

    for valor in serie.dropna():

        # Data
        if isinstance(valor, (pd.Timestamp, datetime, date)):
            qtd_data += 1

            if len(exemplos_data) < 5:
                exemplos_data.append(valor)

            continue

        valor_teste = valor

        # Texto
        if isinstance(valor, str):

            valor_teste = valor.strip()

            # %
            if '%' in valor_teste:
                qtd_percentual += 1

                if len(exemplos_percentual) < 5:
                    exemplos_percentual.append(valor)

                valor_teste = valor_teste.replace('%', '')

            valor_teste = valor_teste.replace(',', '.')

        # Tenta converter para número
        numero = pd.to_numeric(valor_teste, errors='coerce')

        # Texto/caractere que não virou número
        if pd.isna(numero):

            qtd_texto_invalido += 1

            if len(exemplos_texto) < 5:
                exemplos_texto.append(valor)

        else:

            # Número maior que 100
            if numero > 100:

                qtd_maior_100 += 1

                if len(exemplos_maior_100) < 5:
                    exemplos_maior_100.append(valor)

    return {
        'Datas': qtd_data,
        'Com %': qtd_percentual,
        'Texto inválido': qtd_texto_invalido,
        'Maior que 100': qtd_maior_100,
        'Exemplo data': exemplos_data,
        'Exemplo %': exemplos_percentual,
        'Exemplo texto': exemplos_texto,
        'Exemplo >100': exemplos_maior_100
    }


# ============================================
# EXECUTAR CONFERÊNCIA
# ============================================

resultado_antes = []

for ano, info in bases_metricas.items():

    df = info['df']

    for metrica, coluna in info['colunas'].items():

        # Algumas bases podem não possuir determinada métrica
        if coluna not in df.columns:
            print(f"{ano} - {metrica}: coluna {coluna} não existe.")
            continue

        verificacao = verificar_metrica(df[coluna])

        resultado_antes.append({
            'Ano': ano,
            'Metrica': metrica,
            'Coluna': coluna,
            **verificacao
        })


df_verificacao_antes = pd.DataFrame(resultado_antes)

display(df_verificacao_antes)
```

```python
# ============================================
# FUNÇÃO DE CORREÇÃO
# ============================================

def corrigir_metrica(valor):

    # Nulo continua nulo
    if pd.isna(valor):
        return np.nan

    # Excel interpretou nota como data
    if isinstance(valor, (pd.Timestamp, datetime, date)):

        if valor.year <= 2100:
            numero = float(f"{valor.month}.{valor.day}")
        else:
            numero = float(f"{valor.month}.{valor.year}")

    else:

        # Texto
        if isinstance(valor, str):

            valor = (
                valor
                .strip()
                .replace('%', '')
                .replace(',', '.')
            )

        numero = pd.to_numeric(
            valor,
            errors='coerce'
        )

    # Não foi possível converter
    if pd.isna(numero):
        return np.nan

    # Valor maior que 100
    if numero > 100:
        numero = numero / 100

    return float(numero)


# ============================================
# APLICAR NAS MÉTRICAS
# ============================================

for ano, info in bases_metricas.items():

    df = info['df']

    for metrica, coluna in info['colunas'].items():

        if coluna not in df.columns:
            continue

        df[coluna] = (
            df[coluna]
            .apply(corrigir_metrica)
            .astype(float)
        )
```

```python
# ============================================
# CONFERÊNCIA APÓS A CORREÇÃO
# ============================================

resultado_depois = []

for ano, info in bases_metricas.items():

    df = info['df']

    for metrica, coluna in info['colunas'].items():

        if coluna not in df.columns:
            continue

        serie = df[coluna]

        resultado_depois.append({
            'Ano': ano,
            'Metrica': metrica,
            'Coluna': coluna,
            'Tipo': str(serie.dtype),
            'Qtd nulos': serie.isna().sum(),
            'Menor valor': serie.min(),
            'Maior valor': serie.max(),
            'Qtd > 100': (serie > 100).sum()
        })


df_verificacao_depois = pd.DataFrame(resultado_depois)

display(df_verificacao_depois)
```

```python
from datetime import datetime, date

# ============================================
# CORREÇÃO DO INDE
# ============================================

def corrigir_inde(valor):

    # Valor vazio
    if pd.isna(valor):
        return np.nan

    # Excel interpretou o INDE como data
    if isinstance(valor, (pd.Timestamp, datetime, date)):

        # Exemplo: 8.7 virou uma data -> mês 8 / dia 7
        if valor.year <= 2100:
            return float(f"{valor.month}.{valor.day}")

        # Mantendo a mesma regra utilizada anteriormente
        else:
            return float(f"{valor.month}.{valor.year}")

    # Caso seja texto
    if isinstance(valor, str):
        valor = (
            valor
            .strip()
            .replace(',', '.')
        )

    # Converter normalmente para número
    return pd.to_numeric(valor, errors='coerce')


# ============================================
# APLICAR NOS 5 DATAFRAMES
# ============================================

colunas_inde = [
    (df_passos_2020, 'INDE_2020'),
    (df_passos_2021, 'INDE_2021'),
    (df_passos_2022, 'INDE 22'),
    (df_passos_2023, 'INDE 2023'),
    (df_passos_2024, 'INDE 2024')
]

for df, coluna in colunas_inde:

    df[coluna] = (
        df[coluna]
        .apply(corrigir_inde)
        .astype(float)
    )
```

 # Novo dataframe ALUNO_MATERIA #

```python
import pandas as pd
import numpy as np

def process_subjects_for_year(df, year):
    df_subjects_list = []

    # Determine RA column and extract numeric part
    df_RA = None
    if 'NOME' in df.columns:  # For 2020 and 2021
        df_RA = df['NOME'].astype(str).str.replace('ALUNO-', '', regex=False).astype(int)
    elif 'RA' in df.columns:  # For 2022, 2023, 2024
        if pd.api.types.is_string_dtype(df['RA']):
            df_RA = df['RA'].astype(str).str.replace('RA-', '', regex=False).astype(int)
        else:
            df_RA = df['RA'].astype(int)
    else:
        # If no RA column can be determined, return empty DataFrame for subjects
        return pd.DataFrame(columns=['RA', 'Ano', 'Materia', 'Nota'])

    # Define subject column mappings based on year
    subject_cols_map = {}
    if year == 2022:
        subject_cols_map = {'Matem': 'MAT', 'Portug': 'PORT', 'Inglês': 'ING'}
    elif year in [2023, 2024]:
        subject_cols_map = {'Mat': 'MAT', 'Por': 'PORT', 'Ing': 'ING'}
    # For 2020 and 2021, subject_cols_map remains empty, so no subjects will be processed.

    for original_col, materia_name in subject_cols_map.items():
        if original_col in df.columns:
            temp_df = pd.DataFrame({
                'RA': df_RA,
                'Ano': year,
                'Materia': materia_name,
                'Nota': pd.to_numeric(df[original_col], errors='coerce')
            })
            # Drop rows where 'Nota' is NaN (after coerce, these are invalid subject values)
            temp_df.dropna(subset=['Nota'], inplace=True)
            df_subjects_list.append(temp_df)

    return pd.concat(df_subjects_list, ignore_index=True) if df_subjects_list else pd.DataFrame(columns=['RA', 'Ano', 'Materia', 'Nota'])

# Process each year's DataFrame for subjects
aluno_materia_2020 = process_subjects_for_year(df_passos_2020.copy(), 2020)
aluno_materia_2021 = process_subjects_for_year(df_passos_2021.copy(), 2021)
aluno_materia_2022 = process_subjects_for_year(df_passos_2022.copy(), 2022)
aluno_materia_2023 = process_subjects_for_year(df_passos_2023.copy(), 2023)
aluno_materia_2024 = process_subjects_for_year(df_passos_2024.copy(), 2024)

# Concatenate all yearly DataFrames into the final ALUNO_MATERIA DataFrame
ALUNO_MATERIA = pd.concat([
    aluno_materia_2020,
    aluno_materia_2021,
    aluno_materia_2022,
    aluno_materia_2023,
    aluno_materia_2024
], ignore_index=True)

# Ensure RA is integer type if it's not already due to empty dataframes
ALUNO_MATERIA['RA'] = ALUNO_MATERIA['RA'].astype(int)

display(ALUNO_MATERIA.head())
```

# Novo dataframe ALUNO_METRICA #

```python
import pandas as pd
import numpy as np

def process_metrics_for_year(df, year):
    df_metrics_list = []

    # Handle RA extraction based on year/dataframe structure
    df_RA = None
    if 'NOME' in df.columns: # This covers 2020 and 2021 where RA is derived from NOME
        # Ensure 'NOME' values are strings before applying .str.replace
        df_RA = df['NOME'].astype(str).str.replace('ALUNO-', '', regex=False).astype(int)
    elif 'RA' in df.columns: # This covers 2022, 2023, 2024 where RA is a direct column
        if pd.api.types.is_string_dtype(df['RA']):
            df_RA = df['RA'].astype(str).str.replace('RA-', '', regex=False).astype(int)
        else:
            df_RA = df['RA'].astype(int)
    else:
        print(f"Error: Could not find 'RA' or 'NOME' column for RA extraction in year {year}. Returning empty DataFrame.")
        return pd.DataFrame(columns=['RA', 'Ano', 'Metrica', 'Nota', 'Destaque'])

    # Define metric and corresponding Destaque column names based on the year
    metric_col_name_map = {} # Maps standard metric name (e.g., 'IAA') to actual column in df (e.g., 'IAA_2020' or 'IAA')
    destaque_col_name_map = {} # Maps standard metric name (e.g., 'IAA') to actual Destaque column in df (e.g., 'Destaque_IAA_2020' or 'Destaque IAA')

    if year == 2020:
        metric_col_name_map = {
            'IAA': 'IAA_2020', 'IEG': 'IEG_2020', 'IPS': 'IPS_2020',
            'IDA': 'IDA_2020', 'IPP': 'IPP_2020', 'IPV': 'IPV_2020',
            'IAN': 'IAN_2020'
        }
        # Only include Destaque columns specifically requested for 2020
        destaque_col_name_map = {
            'IEG': 'DESTAQUE_IEG_2020',
            'IDA': 'DESTAQUE_IDA_2020',
            'IPV': 'DESTAQUE_IPV_2020'
        }
    elif year == 2021:
        metric_col_name_map = {
            'IAA': 'IAA_2021', 'IEG': 'IEG_2021', 'IPS': 'IPS_2021',
            'IDA': 'IDA_2021', 'IPP': 'IPP_2021', 'IPV': 'IPV_2021',
            'IAN': 'IAN_2021'
        }
        # Only include Destaque columns specifically requested for 2021
        destaque_col_name_map = {
            'IEG': 'DESTAQUE_IEG_2021',
            'IDA': 'DESTAQUE_IDA_2021',
            'IPV': 'DESTAQUE_IPV_2021'
        }
    else: # For 2022, 2023, 2024 - use existing mappings
        metric_col_name_map = {
            'IAA': 'IAA', 'IEG': 'IEG', 'IPP': 'IPP',
            'IPV': 'IPV', 'IPS': 'IPS', 'IDA': 'IDA',
            'IAN': 'IAN'
        }
        destaque_col_name_map = {
            'IAA': 'Destaque IAA', 'IEG': 'Destaque IEG', 'IPP': 'Destaque IPP',
            'IPV': 'Destaque IPV', 'IPS': 'Destaque IPS', 'IDA': 'Destaque IDA',
            'IAN': 'Destaque IAN'
        }

    for standard_metric_name, actual_metric_col in metric_col_name_map.items():
        if actual_metric_col in df.columns:
            temp_metric_df = pd.DataFrame({
                'RA': df_RA,
                'Ano': year,
                'Metrica': standard_metric_name,
                'Nota': pd.to_numeric(df[actual_metric_col], errors='coerce')
            })

            # Initialize 'Destaque' column with pd.NA
            temp_metric_df['Destaque'] = pd.NA

            destaque_col_name_for_metric = destaque_col_name_map.get(standard_metric_name)

            # Check if the specific Destaque column exists for the metric and is not entirely null
            if destaque_col_name_for_metric and destaque_col_name_for_metric in df.columns:
                # Only assign if the column is not entirely null
                if not df[destaque_col_name_for_metric].isnull().all():
                    temp_metric_df['Destaque'] = df[destaque_col_name_for_metric]


            # Drop rows where 'Nota' is NaN (after coerce, these are invalid metric values)
            temp_metric_df.dropna(subset=['Nota'], inplace=True)
            df_metrics_list.append(temp_metric_df)

    return pd.concat(df_metrics_list, ignore_index=True) if df_metrics_list else pd.DataFrame(columns=['RA', 'Ano', 'Metrica', 'Nota', 'Destaque'])


# Process all df_passos dataframes
aluno_metrica_2020 = process_metrics_for_year(df_passos_2020.copy(), 2020)
aluno_metrica_2021 = process_metrics_for_year(df_passos_2021.copy(), 2021)
aluno_metrica_2022 = process_metrics_for_year(df_passos_2022.copy(), 2022)
aluno_metrica_2023 = process_metrics_for_year(df_passos_2023.copy(), 2023)
aluno_metrica_2024 = process_metrics_for_year(df_passos_2024.copy(), 2024)

# Concatenate all yearly DataFrames into the final ALUNO_METRICA DataFrame
ALUNO_METRICA = pd.concat([
    aluno_metrica_2020,
    aluno_metrica_2021,
    aluno_metrica_2022,
    aluno_metrica_2023,
    aluno_metrica_2024
], ignore_index=True)

# Display the first few rows of the updated ALUNO_METRICA DataFrame
display(ALUNO_METRICA.head())
```

# Novo dataframe ALUNO #

```python
def process_aluno_data(df, year):
    temp_aluno_df = pd.DataFrame()

    # RA column extraction and standardization
    if year in [2022, 2023, 2024]:
        # For 2022, 2023, 2024: 'RA' column (e.g., 'RA-1' -> 1)
        temp_aluno_df['RA'] = df['RA'].astype(str).str.replace('RA-', '', regex=False).astype(int)
    elif year in [2020, 2021]:
        # For 2020, 2021: 'NOME' column (e.g., 'ALUNO-1' -> 1)
        temp_aluno_df['RA'] = df['NOME'].astype(str).str.replace('ALUNO-', '', regex=False).astype(int)

    # Nome_aluno column extraction
    if year == 2022:
        temp_aluno_df['Nome_aluno'] = df['Nome']
    elif year in [2023, 2024]:
        temp_aluno_df['Nome_aluno'] = df['Nome Anonimizado']
    elif year in [2020, 2021]:
        temp_aluno_df['Nome_aluno'] = df['NOME']

    # Nascimento column extraction
    if year == 2022:
        temp_aluno_df['Nascimento'] = df['Ano nasc']
    elif year in [2023, 2024]:
        temp_aluno_df['Nascimento'] = df['Data de Nasc']
    else:
        # No direct birth date column for 2020 and 2021 in original dfs
        temp_aluno_df['Nascimento'] = pd.NaT # Not a Time

    # Genero column extraction and standardization
    if 'Gênero' in df.columns:
        temp_aluno_df['Genero'] = df['Gênero']
        temp_aluno_df['Genero'] = temp_aluno_df['Genero'].replace({'Menina': 'Feminino', 'Menino': 'Masculino'})
    else:
        # No 'Gênero' column for 2020 and 2021
        temp_aluno_df['Genero'] = pd.NA

    # Add the year column for sorting and aggregation
    temp_aluno_df['Ano'] = year

    return temp_aluno_df

# Process each year's DataFrame
aluno_2020 = process_aluno_data(df_passos_2020.copy(), 2020)
aluno_2021 = process_aluno_data(df_passos_2021.copy(), 2021)
aluno_2022 = process_aluno_data(df_passos_2022.copy(), 2022)
aluno_2023 = process_aluno_data(df_passos_2023.copy(), 2023)
aluno_2024 = process_aluno_data(df_passos_2024.copy(), 2024)

# Concatenate all yearly DataFrames into a temporary DataFrame
ALUNO_temp = pd.concat([aluno_2020, aluno_2021, aluno_2022, aluno_2023, aluno_2024], ignore_index=True)

# Sort by RA and Year to ensure proper aggregation order
# For Nome_aluno, we want the *first* non-null (e.g., 'ALUNO-1' over 'Aluno-1')
# For Nascimento and Genero, we want the *last* non-null (most recent/complete)
ALUNO_temp = ALUNO_temp.sort_values(by=['RA', 'Ano'])

# Group by RA and aggregate to get a single, most complete record per student
ALUNO = ALUNO_temp.groupby('RA').agg({
    'Nome_aluno': lambda x: x.dropna().iloc[0] if not x.dropna().empty else pd.NA,
    'Nascimento': lambda x: x.dropna().iloc[-1] if not x.dropna().empty else pd.NaT,
    'Genero': lambda x: x.dropna().iloc[-1] if not x.dropna().empty else pd.NA
}).reset_index()

# Ensure 'RA' column is of integer type
ALUNO['RA'] = ALUNO['RA'].astype(int)

# Convert 'Nascimento' to datetime if not already, then format to 'dd/mm/yyyy'
ALUNO['Nascimento'] = pd.to_datetime(ALUNO['Nascimento'], errors='coerce')
ALUNO['Nascimento'] = ALUNO['Nascimento'].dt.strftime('%d/%m/%Y')

# Display the first few rows of the ALUNO DataFrame
display(ALUNO.head())
```

```python
def process_passos_data(df, year):
    temp_passos_df = pd.DataFrame()

    # --- RA (Nr inteiro) ---
    if year in [2020, 2021]:
        # For 2020, 2021: 'NOME' column (e.g., 'ALUNO-1' -> 1)
        temp_passos_df['RA'] = df.get('NOME', pd.Series(index=df.index, dtype=object)).astype(str).str.replace('ALUNO-', '', regex=False).astype(int)
    elif year in [2022, 2023, 2024]:
        # For 2022, 2023, 2024: 'RA' column (e.g., 'RA-1' -> 1)
        temp_passos_df['RA'] = df.get('RA', pd.Series(index=df.index, dtype=object)).astype(str).str.replace('RA-', '', regex=False).astype(int)

    # --- Ano (Nr inteiro) ---
    temp_passos_df['Ano'] = year

    # --- INDE (decimal) ---
    inde_col = None
    if year == 2020:
        inde_col = 'INDE_2020'
    elif year == 2021:
        inde_col = 'INDE_2021'
    elif year == 2022:
        inde_col = 'INDE 22'
    elif year == 2023:
        inde_col = 'INDE 2023'
    elif year == 2024:
        inde_col = 'INDE 2024'
    temp_passos_df['INDE'] = pd.to_numeric(df.get(inde_col, pd.Series(index=df.index, dtype=object)), errors='coerce')


    # --- Fase (Nr inteiro) ---
    if year == 2020:
        # Ex.: "2H" -> 2
        fase_raw = df.get(
            'FASE_TURMA_2020',
            pd.Series(index=df.index, dtype=object)
        ).astype(str)

        fase_series = fase_raw.str.extract(r'(\d+)', expand=False)

        temp_passos_df['Fase'] = (
            pd.to_numeric(fase_series, errors='coerce')
            .astype('Int64')
        )

    elif year == 2021:
        # Ex.: 2.0 -> 2
        temp_passos_df['Fase'] = (
            pd.to_numeric(
                df.get(
                    'FASE_2021',
                    pd.Series(index=df.index, dtype=object)
                ),
                errors='coerce'
            )
            .astype('Int64')
        )

    elif year == 2022:
        # Mantém o valor numérico da coluna Fase
        temp_passos_df['Fase'] = (
            pd.to_numeric(
                df.get(
                    'Fase',
                    pd.Series(index=df.index, dtype=object)
                ),
                errors='coerce'
            )
            .astype('Int64')
        )

    elif year == 2023:
        # Regras:
        # "FASE 1" -> 1
        # "FASE 2" -> 2
        # "ALFA"   -> 0
        # Nulo     -> nulo

        fase_raw = (
            df['Fase']
            .astype('string')
            .str.strip()
        )

        # Começa preservando os nulos da origem
        fase_tratada = pd.Series(
            pd.NA,
            index=df.index,
            dtype='Int64'
        )

        # ALFA -> 0
        mascara_alfa = fase_raw.str.upper().eq('ALFA')
        fase_tratada.loc[mascara_alfa] = 0

        # FASE 1, FASE 2 etc. -> extrai somente o número
        mascara_fase = fase_raw.str.upper().str.startswith('FASE', na=False)

        numeros = (
            fase_raw.loc[mascara_fase]
            .str.extract(r'(\d+)', expand=False)
        )

        fase_tratada.loc[mascara_fase] = pd.to_numeric(
            numeros,
            errors='coerce'
        ).astype('Int64')

        temp_passos_df['Fase'] = fase_tratada


    elif year == 2024:
        # Regras:
        # ALFA -> 0
        # 1A   -> 1
        # 2H   -> 2
        # 9    -> 9
        # nulo -> nulo

        fase_raw = (
            df['Fase']
            .astype('string')
            .str.strip()
            .str.upper()
        )

        # Extrai somente os números
        fase_tratada = fase_raw.str.extract(
            r'(\d+)',
            expand=False
        )

        # ALFA = 0
        fase_tratada.loc[
            fase_raw.eq('ALFA')
        ] = '0'

        temp_passos_df['Fase'] = (
            pd.to_numeric(
                fase_tratada,
                errors='coerce'
            )
            .astype('Int64')
        )


    # --- Turma (caracter) ---
    if year == 2020:
        # Ex.: "2H" -> "H"
        turma_raw = df.get(
            'FASE_TURMA_2020',
            pd.Series(index=df.index, dtype=object)
        ).astype('string').str.strip()

        temp_passos_df['Turma'] = (
            turma_raw
            .str.extract(r'^\s*\d*\s*([A-Za-z])', expand=False)
            .str.upper()
        )

    elif year == 2021:
        # Copia TURMA_2021
        temp_passos_df['Turma'] = (
            df.get(
                'TURMA_2021',
                pd.Series(index=df.index, dtype=object)
            )
            .astype('string')
            .str.strip()
            .replace('', pd.NA)
        )

    elif year == 2022:
        # Copia Turma sem alteração
        temp_passos_df['Turma'] = (
            df.get(
                'Turma',
                pd.Series(index=df.index, dtype=object)
            )
            .astype('string')
            .str.strip()
            .replace('', pd.NA)
        )

    elif year in [2023, 2024]:
        turma_raw = df.get(
            'Turma',
            pd.Series(index=df.index, dtype=object)
        ).astype('string').str.strip()

        # Caso especial:
        # "ALFA B - G2/G3" -> "B"
        turma_alfa = turma_raw.str.extract(
            r'(?i)\bALFA\s*([A-Za-z])',
            expand=False
        )

        # Casos normais:
        # "2H" -> "H"
        # "3A" -> "A"
        # "B"  -> "B"
        turma_normal = turma_raw.str.extract(
            r'^\s*\d*\s*([A-Za-z])',
            expand=False
        )

        temp_passos_df['Turma'] = (
            turma_alfa
            .fillna(turma_normal)
            .str.upper()
        )

    # --- Idade (Nr inteiro) ---
    idade_col = None
    if year == 2020:
        idade_col = 'IDADE_ALUNO_2020'
    elif year == 2022:
        idade_col = 'Idade 22'
    elif year in [2023, 2024]:
        idade_col = 'Idade'
    temp_passos_df['Idade'] = pd.to_numeric(df.get(idade_col, pd.Series(index=df.index, dtype=object)), errors='coerce').astype('Int64')

    # --- Ano_ingresso (Ano) ---
    ano_ingresso_col = 'Ano ingresso'
    temp_passos_df['Ano_ingresso'] = pd.to_numeric(df.get(ano_ingresso_col, pd.Series(index=df.index, dtype=object)), errors='coerce').astype('Int64')

    # --- Instituicao_ensino (caracter) ---
    inst_ensino_col = None
    if year == 2020:
        inst_ensino_col = 'INSTITUICAO_ENSINO_ALUNO_2020'
    elif year == 2021:
        inst_ensino_col = 'INSTITUICAO_ENSINO_ALUNO_2021'
    elif year in [2022, 2023, 2024]:
        inst_ensino_col = 'Instituição de ensino'
    temp_passos_df['Instituicao_ensino'] = df.get(inst_ensino_col, pd.Series(index=df.index, dtype=object)).astype(str).replace({'nan': pd.NA})

    # --- CG, CF, CT (decimal) ---
    temp_passos_df['CG'] = pd.to_numeric(df.get('Cg', pd.Series(index=df.index, dtype=object)), errors='coerce')
    temp_passos_df['CF'] = pd.to_numeric(df.get('Cf', pd.Series(index=df.index, dtype=object)), errors='coerce')
    temp_passos_df['CT'] = pd.to_numeric(df.get('Ct', pd.Series(index=df.index, dtype=object)), errors='coerce')

    # --- Nr_avaliacoes (Nr inteiro) ---
    nr_av_col = 'Nº Av'
    temp_passos_df['Nr_avaliacoes'] = pd.to_numeric(df.get(nr_av_col, pd.Series(index=df.index, dtype=object)), errors='coerce').astype('Int64')

    # --- Rec_psicologia (caracter) ---
    rec_psicologia_col = 'Rec Psicologia'
    temp_passos_df['Rec_psicologia'] = df.get(rec_psicologia_col, pd.Series(index=df.index, dtype=object)).astype(str).replace({'nan': pd.NA})

    # --- Indicado_bolsa (caracter) ---
    indicado_col = 'Indicado'
    temp_passos_df['Indicado_bolsa'] = df.get(indicado_col, pd.Series(index=df.index, dtype=object)).astype(str).replace({'nan': pd.NA})

    # --- Atingiu_PV (caracter) ---
    atingiu_pv_col = 'Atingiu PV'
    temp_passos_df['Atingiu_PV'] = df.get(atingiu_pv_col, pd.Series(index=df.index, dtype=object)).astype(str).replace({'nan': pd.NA})

    # --- Fase_ideal (Nr inteiro) ---
    if year == 2021:
        fase_ideal_series_raw = df.get('NIVEL_IDEAL_2021', pd.Series(index=df.index, dtype=object)).astype(str)
        fase_ideal_series = fase_ideal_series_raw.str.extract(r'(\d+)', expand=False)
        # Removed .fillna('0') to preserve actual NaNs
        fase_ideal_series[fase_ideal_series_raw.str.contains('ALFA', na=False, case=False)] = '0'
        temp_passos_df['Fase_ideal'] = pd.to_numeric(fase_ideal_series, errors='coerce').astype('Int64')
    elif year in [2022, 2023, 2024]:
        fase_ideal_col_name = 'Fase Ideal' if 'Fase Ideal' in df.columns else 'Fase ideal'
        fase_ideal_series_raw = df.get(fase_ideal_col_name, pd.Series(index=df.index, dtype=object)).astype(str)
        fase_ideal_series = fase_ideal_series_raw.str.extract(r'(\d+)', expand=False)
        # Removed .fillna('0') to preserve actual NaNs
        fase_ideal_series[fase_ideal_series_raw.str.contains('ALFA', na=False, case=False)] = '0'
        temp_passos_df['Fase_ideal'] = pd.to_numeric(fase_ideal_series, errors='coerce').astype('Int64')
    else: # 2020 has no direct Fase_ideal
        temp_passos_df['Fase_ideal'] = pd.NA

    # --- Desafagem (Nr inteiro) ---
    desafagem_col = None
    if year == 2021:
        desafagem_col = 'DEFASAGEM_2021'
    elif year == 2022:
        desafagem_col = 'Defas'
    elif year in [2023, 2024]:
        desafagem_col = 'Defasagem'
    temp_passos_df['Desafagem'] = pd.to_numeric(df.get(desafagem_col, pd.Series(index=df.index, dtype=object)), errors='coerce').astype('Int64')

    # --- Escola (caracter) ---
    escola_col = 'Escola'
    temp_passos_df['Escola'] = df.get(escola_col, pd.Series(index=df.index, dtype=object)).astype(str).replace({'nan': pd.NA})

    # --- Status (caracter) ---
    status_col = 'Ativo/ Inativo'
    temp_passos_df['Status'] = df.get(status_col, pd.Series(index=df.index, dtype=object)).astype(str).replace({'nan': pd.NA})

    # Ensure all target columns exist, filling with NA if they were not populated for a specific year
    target_columns = ['RA', 'Ano', 'INDE', 'Fase', 'Turma', 'Idade', 'Ano_ingresso', 'Instituicao_ensino',
                      'CG', 'CF', 'CT', 'Nr_avaliacoes', 'Rec_psicologia', 'Indicado_bolsa',
                      'Atingiu_PV', 'Fase_ideal', 'Desafagem', 'Escola', 'Status']
    for col in target_columns:
        if col not in temp_passos_df.columns:
            temp_passos_df[col] = pd.NA

    # Reorder columns to match the requested order
    temp_passos_df = temp_passos_df[target_columns]

    return temp_passos_df

# Process each year's DataFrame
aluno_passos_2020 = process_passos_data(df_passos_2020.copy(), 2020)
aluno_passos_2021 = process_passos_data(df_passos_2021.copy(), 2021)
aluno_passos_2022 = process_passos_data(df_passos_2022.copy(), 2022)
aluno_passos_2023 = process_passos_data(df_passos_2023.copy(), 2023)
aluno_passos_2024 = process_passos_data(df_passos_2024.copy(), 2024)

# Concatenate all yearly DataFrames into the final ALUNO_PASSOS DataFrame
ALUNO_PASSOS = pd.concat([aluno_passos_2020, aluno_passos_2021, aluno_passos_2022, aluno_passos_2023, aluno_passos_2024], ignore_index=True)

# Display the first few rows of the new DataFrame
display(ALUNO_PASSOS.head())
```

### Preenchendo os INDE faltantes

```python
# ============================================================
# PREENCHER INDE AUSENTE A PARTIR DAS MÉTRICAS
# Chave: RA + Ano
# ============================================================

# Garantir que INDE vazio seja tratado como nulo
ALUNO_PASSOS['INDE'] = pd.to_numeric(
    ALUNO_PASSOS['INDE'],
    errors='coerce'
)

# ------------------------------------------------------------
# Transformar ALUNO_METRICA:
# uma linha por RA/Ano e uma coluna para cada métrica
# ------------------------------------------------------------

metricas_inde = (
    ALUNO_METRICA[
        ALUNO_METRICA['Metrica'].isin(
            ['IAN', 'IDA', 'IEG', 'IAA', 'IPS', 'IPP', 'IPV']
        )
    ]
    .pivot_table(
        index=['RA', 'Ano'],
        columns='Metrica',
        values='Nota',
        aggfunc='first'
    )
    .reset_index()
)

# ------------------------------------------------------------
# Criar base auxiliar mantendo o índice da ALUNO_PASSOS
# ------------------------------------------------------------

base_calculo = ALUNO_PASSOS[
    ['RA', 'Ano', 'Fase', 'INDE']
].copy()

base_calculo['_indice'] = base_calculo.index

base_calculo = base_calculo.merge(
    metricas_inde,
    on=['RA', 'Ano'],
    how='left'
)

# Coluna para armazenar somente os INDEs que serão calculados
base_calculo['INDE_calculado'] = np.nan


# ============================================================
# FASES 0 A 7
#
# INDE =
# IAN * 0.1 +
# IDA * 0.2 +
# IEG * 0.2 +
# IAA * 0.1 +
# IPS * 0.1 +
# IPP * 0.1 +
# IPV * 0.2
# ============================================================

metricas_fase_0_7 = [
    'IAN',
    'IDA',
    'IEG',
    'IAA',
    'IPS',
    'IPP',
    'IPV'
]

tem_metricas_0_7 = (
    base_calculo[metricas_fase_0_7]
    .notna()
    .all(axis=1)
)

mask_0_7 = (
    base_calculo['INDE'].isna()
    & base_calculo['Fase'].notna()
    & base_calculo['Fase'].between(0, 7)
    & tem_metricas_0_7
)

base_calculo.loc[mask_0_7, 'INDE_calculado'] = (
    base_calculo.loc[mask_0_7, 'IAN'] * 0.1
    + base_calculo.loc[mask_0_7, 'IDA'] * 0.2
    + base_calculo.loc[mask_0_7, 'IEG'] * 0.2
    + base_calculo.loc[mask_0_7, 'IAA'] * 0.1
    + base_calculo.loc[mask_0_7, 'IPS'] * 0.1
    + base_calculo.loc[mask_0_7, 'IPP'] * 0.1
    + base_calculo.loc[mask_0_7, 'IPV'] * 0.2
)


# ============================================================
# FASE 8
#
# INDE =
# IAN * 0.1 +
# IDA * 0.4 +
# IEG * 0.2 +
# IAA * 0.1 +
# IPS * 0.2
# ============================================================

metricas_fase_8 = [
    'IAN',
    'IDA',
    'IEG',
    'IAA',
    'IPS'
]

tem_metricas_8 = (
    base_calculo[metricas_fase_8]
    .notna()
    .all(axis=1)
)

mask_8 = (
    base_calculo['INDE'].isna()
    & base_calculo['Fase'].eq(8)
    & tem_metricas_8
)

base_calculo.loc[mask_8, 'INDE_calculado'] = (
    base_calculo.loc[mask_8, 'IAN'] * 0.1
    + base_calculo.loc[mask_8, 'IDA'] * 0.4
    + base_calculo.loc[mask_8, 'IEG'] * 0.2
    + base_calculo.loc[mask_8, 'IAA'] * 0.1
    + base_calculo.loc[mask_8, 'IPS'] * 0.2
)


# ============================================================
# LEVAR O INDE CALCULADO PARA ALUNO_PASSOS
# ============================================================

inde_calculado = (
    base_calculo
    .set_index('_indice')['INDE_calculado']
)

mask_preencher = (
    ALUNO_PASSOS['INDE'].isna()
    & inde_calculado.notna()
)

ALUNO_PASSOS.loc[
    mask_preencher,
    'INDE'
] = inde_calculado.loc[mask_preencher]

# Arredondar para 3 casas decimais
ALUNO_PASSOS['INDE'] = ALUNO_PASSOS['INDE'].round(3)
```
